```python
import numpy as np
import pandas as pd
import tensorflow as tf
from keras import Sequential
from keras.layers import Dense
```

```python
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,11,7]], columns = ['cgpa','profile_score', 'lpa'])
```

```python
df
```

```python
model = Sequential()
model.add(Dense(2,activation= 'linear', input_dim = 2))
model.add(Dense(1, activation = 'linear'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weights = [
    np.array([[0.1, 0.1], [0.1, 0.1]], dtype=np.float32),  # shape (input_dim, units) = (2,2)
    np.array([0., 0.], dtype=np.float32),                    # bias for first Dense layer
    np.array([[0.1], [0.1]], dtype=np.float32),              # shape (units, output_units) = (2,1)
    np.array([0.], dtype=np.float32)                         # bias for second Dense layer
]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error', optimizer=optimizer)
```

```python
X = df.iloc[:,0:-1].values
y = df.iloc[:,-1].values
model.fit(X, y, epochs=100, verbose=1, batch_size=1)
```

```python
new_sample = np.array([[8.5, 8.5]], dtype=np.float32)  
pred = model.predict(new_sample)
print("Prediction for new sample:", pred[0,0])
```

```python
for inp, true_val, pred_val in zip(X, y, predictions):
    print(f"Input={inp}, True={true_val}, Pred={pred_val[0]:.4f}")
```

```python
from sklearn.metrics import r2_score

predictions = model.predict(X)
print("MSE loss:", np.mean((y - predictions.flatten()) ** 2))
print("R2 score:", r2_score(y, predictions.flatten()))

for inp, true_val, pred_val in zip(X, y, predictions):
    print(f"Input={inp}, True={true_val}, Pred={pred_val[0]:.4f}")
```
