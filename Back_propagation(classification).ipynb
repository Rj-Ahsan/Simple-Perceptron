```python
import numpy as np
import pandas as pd
import tensorflow as tf
from keras import Sequential
from keras.layers import Dense
```

```python
df = pd.DataFrame([[8,8,1],[7,9,0],[6,10,1],[5,11,0]], columns = ['cgpa','profile_score', 'lpa'])
```

```python
df.head()
```

```python
model = Sequential()
model.add(Dense(2,activation= 'sigmoid', input_dim = 2))
model.add(Dense(1, activation = 'sigmoid'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weights = [
    np.array([[0.1, 0.1], [0.1, 0.1]], dtype=np.float32),  # shape (input_dim, units) = (2,2)
    np.array([0., 0.], dtype=np.float32),                    # bias for first Dense layer
    np.array([[0.1], [0.1]], dtype=np.float32),              # shape (units, output_units) = (2,1)
    np.array([0.], dtype=np.float32)                         # bias for second Dense layer
]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = tf.keras.optimizers.Adam(learning_rate= 0.001)
model.compile(loss = 'binary_crossentropy',optimizer = optimizer)
```

```python
model.fit(df[['cgpa','profile_score']], df['lpa'], epochs = 100 , verbose = 1, batch_size =1)
```
